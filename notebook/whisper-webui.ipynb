```python
if i == argmin_j(distance_to_goal_j)
    R_i ← "goalkeeper"
else:
  For each robot i (running asynchronously):
  Initialize P_i[r] for each role r (equal or based on abilities)
  Set penalty coefficient γ = 0.2
  Loop:
    Sense environment to update local_state (ball, goals, teammates)

    # —— Local pheromone + penalty update ——
    For each role r:
      Δ_i[r] ← evaluateSuitability(role r, local_state)
      n_r ← countNeighborsWithRole(r)       # how many neighbors have chosen r
      penalty(r) ← γ · max(0, n_r − 1)      # no penalty if alone
      P_i[r] ← (1 − ρ) * P_i[r] + Δ_i[r] + η − penalty(r)

    # broadcast updated values
    Broadcast message containing {P_i[r] ∀ r} to neighbors
    Receive P_j[r] from neighbor robots j in range

    # —— Candidate determination ——
    For each role r:
      If P_i[r] is highest among {P_i[r], P_j[r] ∀ neighbors j}:
        role_r_candidate_i ← true
      Else:
        role_r_candidate_i ← false

    # choose the role i is top candidate for
    R_i ← argmax_{r} { P_i[r] : role_r_candidate_i is true }
    Execute role R_i behavior this cycle

    Wait for next cycle (e.g., 100 ms)
```
